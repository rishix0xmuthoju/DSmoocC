# Lab 12: Unsupervised Learning: K-Means and Hierarchical Clustering

**Aim:** Discover groups in unlabeled data using K-Means and Hierarchical Clustering.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

X, _ = make_blobs(n_samples=100, centers=3, cluster_std=1.5, random_state=42)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)
print("First 20 cluster labels:", labels[:20])
print("Cluster centers:\n", kmeans.cluster_centers_)
```

```python
plt.scatter(X[:,0], X[:,1], c=labels)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], marker="X", s=200)
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

linked = linkage(X, method="ward")
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data points")
plt.ylabel("Distance")
plt.show()

hier_labels = fcluster(linked, t=3, criterion="maxclust")
print("Hierarchical cluster labels (first 20):", hier_labels[:20])
```

**Result:** K-Means and Hierarchical Clustering identified groups in unlabeled data.
