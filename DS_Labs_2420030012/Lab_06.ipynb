# Lab 6: Exploratory Data Analysis using Statistical Measures

**Aim:** Calculate mean, median, mode, variance, standard deviation, and summary statistics.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
display(df.head())
print("\nMean:\n", df.mean())
print("\nMedian:\n", df.median())
print("\nMode of first feature:", mode(df.iloc[:, 0]))
print("\nVariance:\n", df.var())
print("\nStandard deviation:\n", df.std())
print("\nSummary statistics:")
display(df.describe())
```

**Result:** Descriptive statistics were calculated for the Iris dataset.
