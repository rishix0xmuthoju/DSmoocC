# Lab 3: Similarity and Dissimilarity Measures

**Aim:** Calculate distance and similarity measures for numerical data and text.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import numpy as np
from scipy.spatial.distance import euclidean, cityblock, minkowski
from scipy.stats import pearsonr, spearmanr

p1 = np.array([1, 2, 3])
p2 = np.array([4, 6, 3])
print("Euclidean distance:", euclidean(p1, p2))
print("Manhattan distance:", cityblock(p1, p2))
print("Minkowski distance (p=3):", minkowski(p1, p2, 3))
print("Pearson correlation:", pearsonr([1,2,3,4], [2,4,5,8]).statistic)
print("Spearman correlation:", spearmanr([1,2,3,4], [2,4,5,8]).statistic)
```

```python
def hamming_distance(a, b):
    if len(a) != len(b):
        raise ValueError("Strings must have equal length")
    return sum(x != y for x, y in zip(a, b))

def jaccard_similarity(a, b):
    a, b = set(a), set(b)
    return len(a & b) / len(a | b)

print("Hamming distance (karolin, kathrin):", hamming_distance("karolin", "kathrin"))
print("Jaccard similarity:", jaccard_similarity("data science".split(), "data mining".split()))
```

**Result:** Euclidean, Manhattan, Minkowski, correlation, Hamming, and Jaccard measures were calculated.
