# Lab 2: Data Cleaning Techniques

**Aim:** Perform missing-value handling, duplicate removal, format correction, and outlier detection.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena", "Ravi", "Kiran"],
    "Age": [20, np.nan, 22, np.nan, 120],
    "City": ["HYD", "hyd", "Chennai", "hyd", "Delhi"],
    "Marks": [80, 75, np.nan, 75, 90]
})
print("Original data:")
display(df)
```

```python
# Fill missing numerical values with median
for col in ["Age", "Marks"]:
    df[col] = df[col].fillna(df[col].median())

# Normalize text formats and remove duplicate rows
df["City"] = df["City"].str.title()
df = df.drop_duplicates().reset_index(drop=True)

# Detect outliers using IQR (Age column)
q1, q3 = df["Age"].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
df["Age_Outlier"] = (df["Age"] < lower) | (df["Age"] > upper)

print("Cleaned data:")
display(df)
print("Outlier limits:", lower, "to", upper)
```

**Result:** Missing values were filled, duplicates removed, text formats standardized, and outliers identified.
