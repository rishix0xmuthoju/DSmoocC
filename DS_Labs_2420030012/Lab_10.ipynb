# Lab 10: Regression Models for Predictive Analytics

**Aim:** Implement Linear Regression for house-price prediction and Logistic Regression for purchase classification.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import numpy as np
from sklearn.linear_model import LinearRegression, LogisticRegression

# Linear Regression: house size -> price (lakhs)
X_linear = np.array([[1000],[1200],[1500],[1800],[2000]])
y_linear = np.array([50,55,65,80,90])
linear_model = LinearRegression().fit(X_linear, y_linear)
predicted_price = linear_model.predict([[1700]])
print("Predicted price for 1700 sq.ft house:", round(predicted_price[0], 2), "lakhs")
```

```python
# Logistic Regression: price -> purchase (0=no, 1=yes)
X_logistic = np.array([[40],[50],[60],[70],[80],[90]])
y_logistic = np.array([0,0,0,1,1,1])
logistic_model = LogisticRegression().fit(X_logistic, y_logistic)
test_price = np.array([[75]])
print("Purchase prediction (0/1):", logistic_model.predict(test_price)[0])
print("Probabilities [not purchased, purchased]:", logistic_model.predict_proba(test_price)[0].round(3))
```

**Result:** Linear Regression predicted a numerical house price and Logistic Regression classified purchase behaviour.
