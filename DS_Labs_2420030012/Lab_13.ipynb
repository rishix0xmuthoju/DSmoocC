# Lab 13: Text Analytics and Movie Recommendation

**Aim:** Preprocess text and build a content-based movie recommendation system using TF-IDF and cosine similarity.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import re
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = "The movie was amazing, exciting, and full of action scenes!"
tokens = re.findall(r"\\b[a-z]+\\b", text.lower())
stop_words = {"the", "was", "and", "of", "a", "an", "is"}
filtered_tokens = [word for word in tokens if word not in stop_words]
print("Original text:", text)
print("Processed tokens:", filtered_tokens)
```

```python
movies = pd.DataFrame({
    "Movie": ["Avengers", "Batman", "Superman", "Titanic", "Inception"],
    "Description": [
        "action superhero marvel fight",
        "dark superhero action detective",
        "superhero flying action hero",
        "romance ship love drama",
        "dream science fiction thriller"
    ]
})
display(movies)

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(movies["Description"])
similarity = cosine_similarity(matrix)
movie_name = "Batman"
movie_index = movies.index[movies["Movie"] == movie_name][0]
scores = list(enumerate(similarity[movie_index]))
scores = sorted(scores, key=lambda item: item[1], reverse=True)
print("Recommended movies for", movie_name)
for index, score in scores[1:4]:
    print(movies.iloc[index]["Movie"], "- similarity:", round(score, 3))
```

**Result:** Text preprocessing and a content-based movie recommendation system were implemented.
