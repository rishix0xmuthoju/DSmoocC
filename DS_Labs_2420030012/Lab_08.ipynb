# Lab 8: Linear Algebra and Matrix Decomposition

**Aim:** Perform matrix operations, LU decomposition, QR decomposition, SVD, and eigenvalue decomposition.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import numpy as np
from scipy.linalg import lu

A = np.array([[4., 3.], [6., 3.]])
B = np.array([[1., 2.], [3., 4.]])
print("A + B:\n", A + B)
print("A x B:\n", A @ B)
print("Transpose of A:\n", A.T)
print("Determinant of A:", np.linalg.det(A))
```

```python
# LU decomposition
P, L, U = lu(A)
print("LU decomposition")
print("P:\n", P, "\nL:\n", L, "\nU:\n", U)
print("Check P@L@U = A:", np.allclose(P @ L @ U, A))

# QR decomposition
Q, R = np.linalg.qr(A)
print("\nQR decomposition")
print("Q:\n", Q, "\nR:\n", R)
print("Check Q@R = A:", np.allclose(Q @ R, A))
```

```python
# Singular Value Decomposition
u, s, vt = np.linalg.svd(A)
print("Singular values:", s)

# Eigenvalues and eigenvectors
values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
```

**Result:** Matrix operations and decomposition methods were successfully demonstrated.
