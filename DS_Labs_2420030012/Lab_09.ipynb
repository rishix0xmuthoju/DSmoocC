# Lab 9: Probability Distributions and Hypothesis Testing

**Aim:** Generate probability distributions and perform Z-test and independent t-test.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
normal_data = np.random.normal(loc=50, scale=10, size=1000)
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)
poisson_data = np.random.poisson(lam=4, size=1000)

plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")
plt.show()
plt.hist(binomial_data, bins=range(12), align="left")
plt.title("Binomial Distribution")
plt.show()
plt.hist(poisson_data, bins=range(12), align="left")
plt.title("Poisson Distribution")
plt.show()
```

```python
# One-sample Z statistic (population standard deviation known)
sample = np.array([78,82,85,88,90,76,95,89,84,91])
population_mean = 80
population_sd = 10
z_stat = (sample.mean() - population_mean) / (population_sd / np.sqrt(len(sample)))
p_value_z = 2 * (1 - stats.norm.cdf(abs(z_stat)))
print("Z statistic:", round(z_stat, 3), "P-value:", round(p_value_z, 4))

# Independent t-test
class_a = np.array([78,82,85,88,90])
class_b = np.array([72,75,70,68,74])
t_stat, p_value_t = stats.ttest_ind(class_a, class_b)
print("t statistic:", round(t_stat, 3), "P-value:", round(p_value_t, 4))
print("Decision (alpha=0.05):", "Reject H0" if p_value_t < 0.05 else "Fail to reject H0")
```

**Result:** Probability distributions and statistical hypothesis tests were implemented.
