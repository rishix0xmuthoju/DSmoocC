# Lab 5: Data Transformation and Dimensionality Reduction

**Aim:** Apply normalization, standardization, log transformation, and PCA.

**How to run:** Open this notebook in Google Colab and select **Runtime → Run all**.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

df = pd.DataFrame({"A":[10,20,30,40,50], "B":[5,15,25,35,45]})
print("Original data:")
display(df)
print("Min-Max normalization:")
display(pd.DataFrame(MinMaxScaler().fit_transform(df), columns=df.columns))
print("Standardization:")
display(pd.DataFrame(StandardScaler().fit_transform(df), columns=df.columns))
```

```python
# Log transformation (positive values only)
print("Log transformed A:", np.log1p(df["A"]).round(3).tolist())

# PCA on Iris dataset
from sklearn.datasets import load_iris
iris = load_iris()
X_pca = StandardScaler().fit_transform(iris.data)
pca = PCA(n_components=2)
reduced = pca.fit_transform(X_pca)
print("Original dimensions:", iris.data.shape[1])
print("Reduced dimensions:", reduced.shape[1])
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

**Result:** Data transformation and PCA dimensionality reduction were performed.
